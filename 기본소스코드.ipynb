### 🗓️ Day 5: 객체 지향 프로그래밍 기초

```python
class Account:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def show(self):
        print(f"{self.name} 잔액: {self.balance}")

account = Account("Hugo", 1000000)
account.deposit(500000)
account.show()
```

### 🗓️ Day 13: 종합 실습 - 데이터 파이프라인

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. 로드
df = pd.read_excel("financial_data_sample.xlsx")

# 2. 정제
df.fillna(0, inplace=True)

# 3. 핵심 지표 추출
summary = df.groupby("연도")["매출"].sum()

# 4. 시각화
summary.plot(kind="bar")
plt.show()

# 5. 저장
summary.to_csv("연도별_매출_보고서.csv", encoding="utf-8-sig")
```

### 🗓️ Day 14: 핵심 문법 정리 & 포트폴리오 완성하기

```python
# 핵심 문법 정리
# 변수, 자료형, 조건문, 반복문, 함수, 클래스, 외부 라이브러리, 데이터 핸들링
# -> 모두 종합된 예제 작성 가능
# 예: 전체 재무 데이터 처리 클래스 정의 후 보고서 생성

class FinancialDataHandler:
    def __init__(self, filepath):
        import pandas as pd
        self.df = pd.read_excel(filepath)

    def summary(self):
        return self.df.describe()

    def save_report(self, filename):
        summary = self.df.describe()
        summary.to_csv(filename, encoding="utf-8-sig")

# 사용 예
handler = FinancialDataHandler("financial_data_sample.xlsx")
print(handler.summary())
handler.save_report("final_report.csv")
```
